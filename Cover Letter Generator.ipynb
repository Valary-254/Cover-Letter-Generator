```python
# Create OpenAI client

from openai import OpenAI
openai = OpenAI()
```

```python
from pypdf import PdfReader
from docx import Document

def load_cv(pdf_path):
    reader = PdfReader(pdf_path)
    text = ""
    for page in reader.pages:
        text += page.extract_text()
    return text

cv_text = load_cv("Valary Odhiambo Lebenslauf_.pdf")
```

```python
# Step 1: Create your prompts

system_prompt2 = "You are a smart assistant"
user_prompt2 = """ I want you to generate a cover letter from a job description. Make it short, clear and fitting to the role
"""

# Step 2: Make the messages list
def messages2(webPage, cvDoc):
    return [{"role": "system", "content": system_prompt2},
            {"role": "user", "content": user_prompt2 + webPage + cvDoc}] 

# Step 3: Call OpenAI
def createCover(jobDesUrl, cvPath):
    webContent = fetch_website_contents(jobDesUrl)
    cvContent = load_cv(cvPath)
    response = openai.chat.completions.create(model = "gpt-4.1-mini", messages = messages2(webContent, cvContent))
    return response.choices[0].message.content 

def displayCover(jobDesUrl, cvPath):
    createdCover = createCover(jobDesUrl, cvPath)
    display(Markdown(createdCover))

# Step 4: print the result
displayCover("https://support.bioconductor.org/p/104770/", "Valary Odhiambo Lebenslauf_.pdf")
```
